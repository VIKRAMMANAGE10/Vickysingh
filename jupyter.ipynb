```python
import yfinance as yf
import pandas as pd
import requests
from bs4 import BeautifulSoup
import plotly.graph_objects as go
from plotly.subplots import make_subplots

```

```python
#Question 1- using python to extract TSLA data

tesla = yf.Ticker("TSLA")
tesla_data = tesla.history(period="max")
tesla_data.reset_index(inplace=True)
tesla_data.head()
```

```python
#question 2- use of webscrapping to get TSLA revenue data

# Step 1: Get the HTML (using Wayback Machine snapshot to avoid the 403 block)
url = "https://web.archive.org/web/20220101000000/https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"
html_data = requests.get(url).text
soup = BeautifulSoup(html_data, "html.parser")

# Step 2: Create empty DataFrame
tesla_revenue = pd.DataFrame(columns=["Date", "Revenue"])

# Step 3: Find the correct table and extract rows
tables = soup.find_all("tbody")

for table in tables:
    first_row = table.find("tr")
    if first_row and "Tesla Quarterly Revenue" in table.find_previous("th").text if table.find_previous("th") else False:
        target_table = table
        break

print(len(tables)) #Inspect what's actually inside those 6 tables Run this to see the raw structure before filtering:

for i, table in enumerate(tables):
    rows = table.find_all("tr")
    print(f"Table {i}: {len(rows)} rows")
    if rows:
        first_row_cols = rows[0].find_all("td")
        print(f"  First row has {len(first_row_cols)} <td> cells")
        print(f"  Content: {[c.text.strip() for c in first_row_cols]}")


```

```python
tesla_revenue = pd.DataFrame(columns=["Date", "Revenue"])

# Use tables[1] specifically — confirmed as the quarterly revenue table
target_table = tables[1]
rows = target_table.find_all("tr")

data = []
for row in rows:
    cols = row.find_all("td")
    if len(cols) == 2:
        date = cols[0].text.strip()
        revenue = cols[1].text.strip()
        data.append({"Date": date, "Revenue": revenue})

tesla_revenue = pd.DataFrame(data)

# Clean the Revenue column
tesla_revenue["Revenue"] = tesla_revenue["Revenue"].str.replace(r"[\$,]", "", regex=True)
tesla_revenue.dropna(inplace=True)
tesla_revenue = tesla_revenue[tesla_revenue["Revenue"] != ""]

print(len(tesla_revenue))
tesla_revenue.tail()
```

```python
#Question 5-tesla graph

def make_graph(stock_data, revenue_data, title):
    stock_data = stock_data.copy()
    revenue_data = revenue_data.copy()

    stock_data["Date"] = pd.to_datetime(stock_data["Date"])
    revenue_data["Date"] = pd.to_datetime(revenue_data["Date"])
    revenue_data["Revenue"] = pd.to_numeric(revenue_data["Revenue"].astype(str).str.replace(r"[$,]", "", regex=True),errors="coerce")

    fig = make_subplots(rows=2,cols=1,shared_xaxes=True,subplot_titles=("Stock Price", "Revenue"))

    fig.add_trace(go.Scatter(x=stock_data["Date"],y=stock_data["Close"],name="Close Price"),row=1,col=1)

    fig.add_trace(go.Bar(x=revenue_data["Date"],y=revenue_data["Revenue"],name="Revenue"),row=2,col=1)

    fig.update_layout(title=title, height=700)
    fig.show()


make_graph(tesla_data, tesla_revenue, "Tesla Stock Price and Revenue")
```
